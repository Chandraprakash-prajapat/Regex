```python
# What is Feature Engineering ?

# => In Feature engineering we basically create more features or reduce features on the basis of our requirement
```

```python
#  Why do we need feature engineering ?

# Data ==> preprocessing ==> Eda ==> Model evaluation  ==> Model training ==> Model selection ==> feature enginnering  & selection

#  => Feature engineering is the process of transforming raw data into meaningful inputs for machine learning algorithms.
#     It is the foundation of a successful AI project:
#     it unlocks model accuracy, prevents overfitting, and ensures algorithms learn from the right signals
```

```python
# Features
# ==> Numerical => Discrete( no. of cards ) , continous (Age , weight, height)
# ==> Categorical => ordinal( Rank in class) , nominal( city , colors)
```

```python
# Encoding ==> the process of converting the categorical data into numeical data

# 1.one hot encoding ( small data )
# 2.get_dummies( one hot encoding by pandas )
# 3.ordinal encoding
# 4.label encoding
```

```python
#  One Hot Encoding =>  One-hot encoding is a preprocessing technique used to convert categorical data into a numerical format
                      #  so that machine learning algorithms can understand and process it.


# One-hot encoding works by representing each category as a binary vector (a sequence of 0s and 1s)
#  where only one position is "hot" (1) and all others are 0.
```

```python
import pandas as pd

data ={
    'Name' : ['Ram','Shyam','Kalu'],
    'City' : ['Noida', 'Jaipur', 'Delhi']
}

df=pd.DataFrame(data)
df
```

```python
# First method using pandas

# df = pd.get_dummies(df, columns=['City'],dtype=int)
# df
```

```python
# Second method using sklearn (Imp for ML pipeline)

from sklearn.preprocessing import OneHotEncoder
en=OneHotEncoder(sparse_output=False)  #create variable
encode_data= en.fit_transform(df[['City']])

data=pd.DataFrame(encode_data,columns=en.get_feature_names_out())
data


```

```python
#  Label Encoder

from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

data ={
    'pass/fail' : ['pass','fail','pass','fail']
}

df=pd.DataFrame(data)
df
```

```python
# Feature scaling
# standardization (Z-Score)
# Normalization (Min max scaling )

import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
data = {
    'Area': [200, 300, 400, 500, 600],
    'Population' : [30000, 50000,70000, 90000, 95000]
}

df=pd.DataFrame(data)
df
```

```python
scaler=MinMaxScaler()
df['Scaled_Area']=scaler.fit_transform(df[['Area']])
df
```

```python
scaler=StandardScaler()
df['Scaled_Area']=scaler.fit_transform(df[['Area']])
df
```

```python
#  apply on multiple column
scaler=StandardScaler()
scaled_data=scaler.fit_transform(df[['Area','Population']])
pd.DataFrame(scaled_data,columns=df.columns)
```

```python
#  How to handle outliers in data

# we generally drop the outliers when we have small number of outliers
# but without discussing with the team we cant drop that entry

# We cap the outliers with boundary value( upper , lower )



# Steps:-
# visualize the data
# verify outlier are genuine or erros
# remove only obvious erros
#  for valid outliers - capping , transformations ,
#                      robust scaling or models
```

```python
# handle outliers with IQR method

import pandas as pd
df=pd.DataFrame({
    'salary':[500, 20000,25000,27000,30000,31000,500000] }
 )

Q1=df['salary'].quantile(0.25)
Q3=df['salary'].quantile(0.75)
IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# df[(df['salary']<lower) | (df['salary']>upper)]

# lower bound se chhoti values lower bound se replace ho jayegi
# upper bound se badi values upper bound se replace ho jayegi
# baaki values same rhegi

df['Remove Outliers']=df['salary'].clip(lower,upper)
df
```
