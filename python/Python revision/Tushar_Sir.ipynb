Arithmetic Operator :

```python
# Arithmetic Operator : use to perform basic mathematical operations
a=10
b=3
print(a+b)  # add
print(a-b)  # subtract
print(a*b)  # multiply
print(a/b)  # / means -> division (float value return)
print(a%b)  # Remeinder
print(a//b) # // means -> floor division (int return)
print(a**b) # exponent(power)
```

comparision operator

```python
# comparision operator ( return boolean value )

x=5
y=10
print(x==y)   #  comapre both value is equal or not
print(x!=y)   # not equal
print(x>y)    # greater >
print(x<y)    # less than <
print(x>=y)   # greater and equal >=
print(x<=y)   # less than and equal <=
```

Logical operator : combine multiple condtions together

```python
#  Logical operator  => use when multiple conditions check ( AND , OR , NOT )
 # c lang. => & , | , ~

a=True
b=False
print(a and b)  # both condition must be correct for TRUE
print(a or b)   # one condition must be correct for TRUE
print(not a)    #
```

```python
a = 10

print(a > 3 and a>=10)    #  in AND all conditions correct => return True otherwise False

print(a >3 and a > 70)

print( a>3 or a>5 and a>15 )   # first specify AND then OR
```

```python
# Bitwise Operators : Used to perform operations on binary numbers.

# &     =>	  AND
# \|    => 	  OR
# ^	    =>    XOR
# ~	    =>    NOT
# <<	  =>    Left shift
# >>	  =>    Right shift
```

```python
a = 5
b = 3

print(5 & 3)

print( 5|3 )

print(5 ^ 3)

print(~5)

print(5 << 1)

print(5 >> 1)

print(5 >> 1)
```

```python
# Right Shift (>>)
# Rule: Bits ko right side shift karta hai.

print(5 >> 1)
```

```python
# Membership  : a value exists in a sequence (list , string etc.)

nums=[1,2,3,4]

print( 2 in nums)
print(20 in nums)
print(5 not in nums)
```

```python
# Identity operators : two variable point to the same object in memory

a=[1,2,3]
b=a
c=[1,2,3]

print(a is b)
print(a is c)
print(a is not c )
```
