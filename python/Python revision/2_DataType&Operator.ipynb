 Bitwise Operators : Used to perform operations on binary numbers.

 &     =>	  AND       =>  Sets bit to 1 if both bits are 1

 \|    => 	  OR        =>  Sets bit to 1 if one of them bit is 1

 ^	    =>    XOR       => 	Sets bit to 1 if bits are different

 ~	    =>    NOT       => 	Inverts all bits => ~n = -(n+1)

 <<	  =>    Left shift  =>  Shifts bits to the left

 (>>)	  =>    Right shift =>  Shifts bits to the right

```python
a = 5
b = 3

print(5 & 3)

print( 5|3 )

print(5 ^ 3)

print(~5)

print(5 << 1)

print(5 >> 1)
```

```python
# Q. What is call by value and call by reference
# 1. Call by Value
# In Call by Value, a copy of the value is passed to the function.

# Changes made inside the function do not affect the original variable.

def change(x):
    x = 20
    print("Inside function:", x)

a = 10
change(a)

print("Outside function:", a)
```

```python
# 2. Call by Reference
# In Call by Reference, a reference to the original object is passed to the function.

# Changes made to a mutable object inside the function can affect the original object.

def change(mylist):
    mylist.append(4)
    print("Inside function:", mylist)

a = [1, 2, 3]
change(a)

print("Outside function:", a)
```

=> Call by Object Reference in Python uses Call by Object Reference, also known as Call by Sharing.

=> It means when we pass a variable to a function, Python passes a reference to the same object.

=>Both the original variable and the function parameter can refer to that object.
